```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df=pd.read_csv("merged_data_cleaned.csv")
```

```python
df.head()
```

```python
df.groupby("Financial_Year")[
    ["NEFT_Transactions_Lakh",
     "RTGS_Transactions_Lakh",
     "IMPS_Transactions_Lakh",
     "UPI_Transactions_Lakh"]
].sum().sort_index(ascending=True)
```

```python
upi_yearly=df.groupby("Financial_Year")["UPI_Transactions_Lakh"].sum().sort_index(ascending=True)
upi_yearly.plot(kind="line",marker="o", figsize=(8,5))
plt.title("Yearly UPI Transcations")
plt.xlabel("Financial Year")
plt.ylabel("UPI Trascation")
plt.tight_layout()
plt.xticks(rotation=30)
plt.grid(True)
plt.show()
```

```python
df.groupby("Bank_Name")["UPI_Transactions_Lakh"].sum().sort_values(ascending=False).head(10)
```

```python
top10_banks_by_upi=df.groupby("Bank_Name")["UPI_Transactions_Lakh"].sum().sort_values(ascending=False).head(10)
top10_banks_by_upi.plot(kind="bar",figsize=(10,5))
plt.title("Top 10 Banks by UPI Transactions")
plt.xlabel("Bank_name")
plt.ylabel("UPI Transactions in lakhs")
plt.tight_layout()
plt.xticks(rotation=90)
plt.grid(False)
plt.show()
```

```python
df.groupby("Bank_Type")[
    ["UPI_Transactions_Lakh", "Mobile_Transactions_Lakh",
     "Internet_Transactions_Lakh"]
].sum()
```

```python
Banks_Types=df.groupby("Bank_Type")[
    ["UPI_Transactions_Lakh", "Mobile_Transactions_Lakh",
     "Internet_Transactions_Lakh"]].sum()
Banks_Types.plot(kind="bar", figsize=(10, 5))
plt.title("Banks_wise_digital_payment")
plt.xlabel("Bank_Type")
plt.ylabel("Digital_payment_Transactions")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
df.groupby("State")["UPI_Transactions_Lakh"].sum().sort_values(ascending=False).head(10)
```

```python
state_upi=df.groupby("State")["UPI_Transactions_Lakh"].sum().sort_values(ascending=False).head(10)
state_upi.plot(kind="bar",figsize=(10,5),color = [
    'blue',
    'orange',
    'green',
    'red',
    'purple',
    'brown',
    'pink',
    'gray',
    'olive',
    'cyan'
])
plt.title("state wise UPI Transactions")
plt.xlabel("States")
plt.ylabel("UPI Transactions_lakhs")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
digital_yearly = df.groupby("Financial_Year")[
    ["UPI_Transactions_Lakh",
     "Mobile_Transactions_Lakh",
     "Internet_Transactions_Lakh"]
].sum().sort_index()

digital_yearly.plot(
    kind="line",
    marker="o",
    figsize=(10, 5)
)

plt.title("Digital Banking Transactions Growth by Financial Year")
plt.xlabel("Financial Year")
plt.ylabel("Transactions (Lakh)")
plt.xticks(rotation=45)
plt.legend(["UPI", "Mobile Banking", "Internet Banking"])
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
bank_digital = (
    df.groupby("Bank_Name")[
        ["UPI_Transactions_Lakh",
         "Mobile_Transactions_Lakh",
         "Internet_Transactions_Lakh"]
    ]
    .sum()
)

top_digital_banks = bank_digital.sum(axis=1).nlargest(10).index
top_digital = bank_digital.loc[top_digital_banks]

top_digital.plot(kind="bar", figsize=(12, 5))

plt.title("Top 10 Banks by Total Digital Banking Transactions")
plt.xlabel("Bank Name")
plt.ylabel("Transactions (Lakh)")
plt.xticks(rotation=45, ha="right")
plt.legend(["UPI", "Mobile Banking", "Internet Banking"])
plt.tight_layout()
plt.show()
```

```python
digital_cols = [
    "NEFT_Transactions_Lakh",
    "RTGS_Transactions_Lakh",
    "IMPS_Transactions_Lakh",
    "UPI_Transactions_Lakh",
    "Mobile_Transactions_Lakh",
    "Internet_Transactions_Lakh"
]

correlation = df[digital_cols].corr()

plt.figure(figsize=(10, 7))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".5f",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Between Digital Banking Transactions")
plt.tight_layout()
plt.show()
```
